```python
print("Rows with at least one -200:", (missing_pattern > 0).sum())
```

```python
df_clean = df.replace(-200, pd.NA)

print("Missing values after conversion:")
print(df_clean.isna().sum())
```

```python

df_clean = df_clean.drop(columns=["NMHC(GT)"])


df_clean = df_clean.copy()


numeric_columns = df_clean.select_dtypes(include="number").columns

for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )

duplicate_count = df_clean.duplicated().sum()


df_clean = df_clean.drop_duplicates()

print("Cleaning completed!")
print("Duplicate rows found:", duplicate_count)
print("Remaining missing values:", df_clean.isna().sum().sum())
print("Final shape:", df_clean.shape)
```

```python
outlier_summary = {}

for column in numeric_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = (
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ).sum()

    outlier_summary[column] = outliers

    
    df_clean[column] = df_clean[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("Outlier treatment completed!\n")

for column, count in outlier_summary.items():
    print(column, ":", count, "outliers handled")
```

```python


for column in numeric_columns:
    df_clean[column] = pd.to_numeric(
        df_clean[column],
        errors="coerce"
    )

# Fill any remaining missing numeric values
for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )

print("Numeric data fixed.")
print("Remaining missing values:", df_clean.isna().sum().sum())
```

```python


for column in df_clean.columns:
    if column not in ["Date", "Time"]:
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        )
        
        df_clean[column] = df_clean[column].fillna(
            df_clean[column].median()
        )

print("Remaining missing values:",
      df_clean.isna().sum().sum())
```

```python
import matplotlib.pyplot as plt

# 1. CO distribution
plt.figure(figsize=(8, 5))
plt.hist(df_clean["CO(GT)"], bins=40)
plt.title("Distribution of Carbon Monoxide (CO)")
plt.xlabel("CO Concentration")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 2. NOx distribution
plt.figure(figsize=(8, 5))
plt.hist(df_clean["NOx(GT)"], bins=40)
plt.title("Distribution of Nitrogen Oxides (NOx)")
plt.xlabel("NOx Concentration")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 3. Temperature distribution
plt.figure(figsize=(8, 5))
plt.hist(df_clean["T"], bins=40)
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 4. Correlation heatmap
correlation = df_clean.select_dtypes(include="number").corr()

plt.figure(figsize=(12, 8))
plt.imshow(correlation, aspect="auto")
plt.colorbar()
plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)
plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)
plt.title("Correlation Matrix of Air Quality Variables")
plt.tight_layout()
plt.show()
```

```python


print("========== FINAL DATASET ==========")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

print("\nRemaining missing values:")
print(df_clean.isna().sum().sum())

print("\nRemaining duplicate rows:")
print(df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)


df_clean.to_csv("air_quality_cleaned.csv", index=False)

print("\nCleaned dataset saved as: air_quality_cleaned.csv")
```
